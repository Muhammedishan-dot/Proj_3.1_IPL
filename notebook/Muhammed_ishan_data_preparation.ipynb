```python
import sqlite3
import pandas as pd
con=sqlite3.connect('../data/raw/ipl.db')

def q(sql):
    return pd.read_sql(sql, con)
```

```python
def run_sql_file(path):
    with open(path, 'r') as f:
        script = f.read()
    con.executescript(script)
    con.commit()
    print('ran',path)
```
